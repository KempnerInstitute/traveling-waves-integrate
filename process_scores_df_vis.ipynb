```python
# Specify the path to your CSV file
csv_file = 'results_scores/scores_test_mnist_tetro.csv'
df = pd.read_csv(csv_file)
df = df.drop(["run_name", "max_iters"], axis=1)
```

```python
# Define the grouping keys
group_keys = ["dataset", "model_type", "num_layers", "readout_type"]

# Define the metrics for which statistics should be computed
metrics = ["acc", "acc_foreground", "iou", "iou_foreground", "loss", "loss_foreground"]

# Group by the specified keys and compute statistics across the "extension" category
grouped_df = df.groupby(group_keys).agg({
    metric: ['mean', 'std'] for metric in metrics
}).reset_index()
```

```python
def plot_df(the_df, fpath, bg_iou, bg_acc):
    sns.set_theme()
    metrics = {'acc' : 'Acc',
                'acc_foreground' : 'FG-Acc',
                'iou' : 'IoU',
                'iou_foreground' : 'FG-IoU',
                'loss' : 'Loss',
                'loss_foreground' : 'FG-Loss'}
    models = {'cornn_model2' : "NWM",
              'conv_recurrent2' : "LSTM"}
    orig_metrics = ['acc', 'iou', 'loss', 'acc_foreground', 'iou_foreground', 'loss_foreground']
    fig, axes = plt.subplots(2, 3, figsize=(20, 10))
    count = 0
    for i in range(len(axes)):
        for j in range(len(axes[i])):
        # Sample DataFrame
            metric = orig_metrics[count]
            data = {'Readout': [readouts[readout] for readout in the_df['readout_type'].values],
                    'Model': [models[model] for model in the_df['model_type'].values],
                    'Metric': the_df[metric, 'mean'].values,
                    'STD': the_df[metric, 'std'].values}
            df = pd.DataFrame(data)
            # Pivot the data
            pivot_df = df.pivot(index='Readout', columns='Model', values='Metric')
            pivot_df_std = df.pivot(index='Readout', columns='Model', values='STD')
            pivot_df.plot.barh(grid=True, ax=axes[i][j], xerr=pivot_df_std, legend=False)
            axes[i][j].set_xlabel(metrics[metric], fontsize=16)
            axes[i][j].set_ylabel('Readout', fontsize=16)
            # Plot background
            if metric == 'iou':
                axes[i][j].axvline(x=bg_iou, color='red', linestyle='--')
                axes[i][j].set_xlim(bg_iou - 0.1, 1.0)
            elif metric == 'acc':
                axes[i][j].axvline(x=bg_acc, color='red', linestyle='--')
                axes[i][j].set_xlim(bg_acc - 0.1, 1.0)
            count += 1
            axes[i][j].tick_params(which="both", labelsize=16)


    #plt.legend(loc='best')
    handles, labels = axes[0][0].get_legend_handles_labels()
    fig.legend(handles, labels, fontsize=16)
    fig.tight_layout()
    #plt.show()
    plt.savefig(fpath)
    plt.close()
```

```python
plot_df(dfg1[5:], "results_scores/bar_mnist.pdf", mnist_bg_iou, mnist_bg_acc)
plot_df(dfg2[5:], "results_scores/bar_tetro.pdf", tetro_bg_iou, tetro_bg_acc)
```

```python
def plot_df2(the_df, fpath, bg_iou, bg_acc):
    sns.set_theme()
    metrics = {'acc' : 'Acc',
               'acc_foreground' : 'FG-Acc',
                'iou' : 'IoU',
                'iou_foreground' : 'FG-IoU',
                'loss' : 'Loss',
                'loss_foreground' : 'FG-Loss'}
    models = {'cornn_model2' : "NWM",
              'conv_recurrent2' : "LSTM",
              'baseline1_flexible' : "CNN"}
    orig_metrics = ['acc', 'iou', 'loss', 'acc_foreground', 'iou_foreground', 'loss_foreground']
    count = 0
    fig, axes = plt.subplots(2, 3, figsize=(20, 10))
    for count1 in range(len(axes)):
        for count2 in range(len(axes[count1])):
            # Sample DataFrame
            metric = orig_metrics[count]
            layers = []
            the_models = []
            means = []
            stds = []
            for i, model in enumerate(the_df['model_type'].values):
                if model == 'cornn_model2' or model == 'conv_recurrent2':
                    if the_df['readout_type'].values[i] == 'linear':
                        layers.append("Linear")
                        means.append(the_df[metric, 'mean'].values[i])
                        stds.append(the_df[metric, 'std'].values[i])
                        the_models.append(models[the_df['model_type'].values[i]])
                else:
                    layers.append(the_df['num_layers'].values[i])
                    means.append(the_df[metric, 'mean'].values[i])
                    stds.append(the_df[metric, 'std'].values[i])
                    the_models.append(models[the_df['model_type'].values[i]])
                
            data = {
                    'Model': the_models,
                    'Arch' : layers,
                    'Metric': means,
                    'STD': stds,
                    }
            
            df = pd.DataFrame(data)
            # Pivot the data
            pivot_df = df.pivot(index='Model', columns='Arch', values='Metric')
            pivot_df_std = df.pivot(index='Model', columns='Arch', values='STD')
            pivot_df.plot.barh(grid=True, ax=axes[count1][count2], xerr=pivot_df_std, legend=False)
            axes[count1][count2].set_xlabel(metrics[metric], fontsize=16)
            axes[count1][count2].set_ylabel('Model', fontsize=16)
            if metric == 'iou':
                axes[count1][count2].axvline(x=bg_iou, color='red', linestyle='--')
                axes[count1][count2].set_xlim(bg_iou - 0.1, 1.0)
            elif metric == 'acc':
                axes[count1][count2].axvline(x=bg_acc, color='red', linestyle='--')
                axes[count1][count2].set_xlim(bg_acc - 0.1, 1.0)
            axes[count1][count2].tick_params(which="both", labelsize=16)
            
            count += 1


    handles, labels = axes[0][0].get_legend_handles_labels()
    fig.legend(handles, labels, fontsize=16)
    fig.tight_layout()
    #plt.show()
    plt.savefig(fpath)
    plt.close()
```

```python
plot_df2(dfg1, "results_scores/bar_mnist2.pdf", mnist_bg_iou, mnist_bg_acc)
plot_df2(dfg2, "results_scores/bar_tetro2.pdf", tetro_bg_iou, tetro_bg_acc)
```
